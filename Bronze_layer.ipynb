```python
from pyspark.sql.functions import *
```

```python
raw_file_path = '/Volumes/workspace/taxi/raw_data/yellow_tripdata_2023-01.parquet'
bronze_table = 'workspace.taxi.bronze_trips'


df = spark.read.parquet(raw_file_path)
```

```python
df.display()
```

```python
df.write.format('delta').mode('append').saveAsTable(bronze_table)

print(f"Bronze ingestion complete. Rows: {spark.table(bronze_table).count()}")
```
